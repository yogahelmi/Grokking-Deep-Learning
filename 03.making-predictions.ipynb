## 1. Executive Summary & Overview
Chapter 3 explores the first component of the machine learning trilogy: **Prediction (Forward Propagation)**. 

Before a network can learn from mistakes, it must possess the mechanical capability to transform inputs into outputs. Andrew Trask systematically constructs forward propagation across five levels of complexity:
1. **Single Input $\rightarrow$ Single Output:** Scalar multiplication.
2. **Multiple Inputs $\rightarrow$ Single Output:** Vector dot product (weighted combination of features).
3. **Single Input $\rightarrow$ Multiple Outputs:** Scalar-vector scaling (distributing a single input across multiple predictions).
4. **Multiple Inputs $\rightarrow$ Multiple Outputs:** Matrix-vector multiplication (full linear transformation).
5. **Predicting on Predictions (Hidden Layers):** Chained matrix transformations ($\mathbf{y} = \mathbf{W}_2 \mathbf{W}_1 \mathbf{x}$).

### Key Chapter Takeaways:
- **Neural Nodes as Information Routers:** A node is simply a container that holds a number. A weight is an amplification / attenuation factor on the wire connecting two nodes.
- **Matrix Multiplication as Batch Feature Extraction:** Passing an input vector through a weight matrix simultaneously calculates multiple independent dot products, evaluating the input against multiple feature templates.

## 2. Theoretical Deep-Dive: Linear Algebra of Forward Propagation

### 2.1 Case 1: Single Input, Single Output
The simplest conceivable artificial neural network connects one input node $x$ to one output node $\hat{y}$ through a single scalar weight $w$:
$$\hat{y} = x \cdot w$$
Here, the weight represents the ratio of expected output per unit of input.

### 2.2 Case 2: Multiple Inputs, Single Output
When multiple features inform a prediction (e.g., number of toes, win record, fan count):
$$\mathbf{x} = \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{bmatrix}, \quad \mathbf{w} = \begin{bmatrix} w_1 \\ w_2 \\ \vdots \\ w_n \end{bmatrix}$$
The prediction is computed as the vector inner product:
$$\hat{y} = \mathbf{x}^T \mathbf{w} = \sum_{i=1}^n x_i w_i = x_1 w_1 + x_2 w_2 + \dots + x_n w_n$$
In linear algebra terms, $\hat{y}$ is a **linear combination** of the input features weighted by the entries of $\mathbf{w}$.

### 2.3 Case 3: Multiple Inputs, Multiple Outputs
When predicting multiple target variables (e.g., probability of injury, win probability, team morale) from multiple input features:
$$\mathbf{x} \in \mathbb{R}^n, \quad \mathbf{\hat{y}} \in \mathbb{R}^m, \quad \mathbf{W} \in \mathbb{R}^{m \times n}$$
$$\mathbf{\hat{y}} = \mathbf{W} \mathbf{x} = \begin{bmatrix} 
w_{11} & w_{12} & \cdots & w_{1n} \\
w_{21} & w_{22} & \cdots & w_{2n} \\
\vdots & \vdots & \ddots & \vdots \\
w_{m1} & w_{m2} & \cdots & w_{mn}
\end{bmatrix} \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{bmatrix}$$

Each row $\mathbf{w}_{j,:}$ of the weight matrix represents the dedicated weight vector for output unit $j$:
$$\hat{y}_j = \mathbf{w}_{j,:} \cdot \mathbf{x}$$

### 2.4 Case 4: Deep Networks (Predicting on Predictions)
In a network with hidden layers:
$$\mathbf{h} = \mathbf{W}_{ih} \mathbf{x}, \quad \mathbf{\hat{y}} = \mathbf{W}_{hp} \mathbf{h} = \mathbf{W}_{hp} (\mathbf{W}_{ih} \mathbf{x})$$
Notice from linear algebra that the product of two matrices is another matrix:
$$\mathbf{W}_{net} = \mathbf{W}_{hp} \mathbf{W}_{ih}$$
Without non-linear activation functions between layers, stacking linear layers collapses into a single linear map. We will see how activation functions break this linearity in Chapter 6!

## 3. Code Reproduction: From Pure Python to NumPy

### 3.1 Network 1: Simplest Network (Single Input $\rightarrow$ Single Output)


```python
# Andrew Trask: Simplest Neural Network
weight = 0.1

def neural_network_scalar(input_val, weight_val):
    prediction = input_val * weight_val
    return prediction

number_of_toes = [8.5, 9.5, 10, 9]
input_data = number_of_toes[0]

pred = neural_network_scalar(input_data, weight)
print(f"Input toes: {input_data}")
print(f"Predicted outcome: {pred:.4f}")
```

### 3.2 Network 2: Multiple Inputs $\rightarrow$ Single Output
We predict whether the team wins based on:
1. `toes`: Average number of toes per player.
2. `wlrec`: Win-loss record percentage.
3. `nfans`: Number of fans in millions.


```python
# Pure Python Implementation
def w_sum(a, b):
    assert(len(a) == len(b))
    output = 0
    for i in range(len(a)):
        output += (a[i] * b[i])
    return output

weights = [0.1, 0.2, 0.0]

def neural_network_multi_in(input_vec, weight_vec):
    prediction = w_sum(input_vec, weight_vec)
    return prediction

toes = [8.5, 9.5, 9.9, 9.0]
wlrec = [0.65, 0.8, 0.8, 0.9]
nfans = [1.2, 1.3, 0.5, 1.0]

# First game of the season
input_game_1 = [toes[0], wlrec[0], nfans[0]]
pred_multi = neural_network_multi_in(input_game_1, weights)
print(f"Inputs: {input_game_1}")
print(f"Weights: {weights}")
print(f"Prediction (Pure Python): {pred_multi:.4f}")
```

### 3.3 Network 2 in NumPy Vectorized Code
In NumPy, `w_sum` is replaced by the highly optimized `np.dot` function:


```python
import numpy as np

weights_np = np.array([0.1, 0.2, 0.0])

def neural_network_numpy(input_vec, weight_vec):
    prediction = input_vec.dot(weight_vec)
    return prediction

input_np = np.array([toes[0], wlrec[0], nfans[0]])
pred_np = neural_network_numpy(input_np, weights_np)
print(f"Prediction (NumPy dot): {pred_np:.4f}")
```

### 3.4 Network 3: Single Input $\rightarrow$ Multiple Outputs
Given a single input (`wlrec` = 0.65), predict three separate outcomes:
1. `hurt?` (probability of player injury)
2. `win?` (probability of team winning)
3. `sad?` (probability of team morale loss)


```python
def ele_mul(number, vector):
    output = [0] * len(vector)
    assert(len(output) == len(vector))
    for i in range(len(vector)):
        output[i] = number * vector[i]
    return output

weights_single_to_multi = [0.3, 0.2, 0.9]

def neural_network_single_to_multi(input_scalar, weight_vec):
    prediction = ele_mul(input_scalar, weight_vec)
    return prediction

wlrec_game_1 = wlrec[0]
pred_3 = neural_network_single_to_multi(wlrec_game_1, weights_single_to_multi)
print(f"Input Win%: {wlrec_game_1}")
print(f"Predictions [Hurt, Win, Sad]: {[round(p, 4) for p in pred_3]}")
```

### 3.5 Network 4: Multiple Inputs $\rightarrow$ Multiple Outputs
Now we connect 3 inputs (`[toes, win%, fans]`) to 3 outputs (`[hurt?, win?, sad?]`) using a $3 \times 3$ weight matrix:


```python
# Pure Python Matrix-Vector Multiplication
def vect_mat_mul(vect, matrix):
    assert(len(vect) == len(matrix))
    output = [0] * len(matrix)
    for i in range(len(matrix)):
        output[i] = w_sum(vect, matrix[i])
    return output

# Weights matrix (rows = output features, cols = input features)
#                 toes  %win  fans
weights_mimo = [ [0.1,  0.1, -0.3],  # hurt?
                 [0.1,  0.2,  0.0],  # win?
                 [0.0,  1.3,  0.1] ] # sad?

def neural_network_mimo(input_vec, weight_mat):
    prediction = vect_mat_mul(input_vec, weight_mat)
    return prediction

pred_mimo = neural_network_mimo(input_game_1, weights_mimo)
print(f"Inputs: {input_game_1}")
print(f"Predictions [hurt, win, sad]: {[round(p, 4) for p in pred_mimo]}")
```

### 3.6 Network 5: Predicting on Predictions (Hidden Layer Network)
We construct our first **Deep** architecture:
$$\text{Input Layer } (\mathbb{R}^3) \xrightarrow{\mathbf{W}_{ih}} \text{Hidden Layer } (\mathbb{R}^3) \xrightarrow{\mathbf{W}_{hp}} \text{Output Layer } (\mathbb{R}^3)$$


```python
# Input to Hidden Weights (3x3)
ih_wgt = np.array([
    [0.1, 0.2, -0.1],   # hid[0]
    [-0.1, 0.1, 0.9],   # hid[1]
    [0.1, 0.4, 0.1]     # hid[2]
])

# Hidden to Output Weights (3x3)
hp_wgt = np.array([
    [0.3, 1.1, -0.3],   # hurt?
    [0.1, 0.2, 0.0],    # win?
    [0.0, 1.3, 0.1]     # sad?
])

def neural_network_deep(input_vec, w_ih, w_hp):
    # Step 1: Forward propagate input to hidden layer
    hid = input_vec.dot(w_ih)
    # Step 2: Forward propagate hidden layer to prediction
    pred = hid.dot(w_hp)
    return hid, pred

input_vec_np = np.array([toes[0], wlrec[0], nfans[0]])
hidden_activations, final_pred = neural_network_deep(input_vec_np, ih_wgt, hp_wgt)

print(f"Input: {input_vec_np}")
print(f"Hidden Layer Activations: {np.round(hidden_activations, 4)}")
print(f"Final Prediction [hurt, win, sad]: {np.round(final_pred, 4)}")
```

## 4. Practical Linear Algebra Primer: Dimensions & Matrix Shapes
A common point of error in neural network engineering is matrix shape incompatibility. The following rules must always be obeyed:
1. Two matrices $\mathbf{A} \in \mathbb{R}^{m \times k}$ and $\mathbf{B} \in \mathbb{R}^{k \times n}$ can be multiplied if and only if the inner dimensions match ($k = k$).
2. The resulting matrix has dimensions $m \times n$.
3. When using row-vectors in NumPy: `x.dot(W)` requires `x.shape[1] == W.shape[0]`.


```python
a_mat = np.zeros((2, 4))
b_mat = np.zeros((4, 3))
c_mat = a_mat.dot(b_mat)

print(f"Shape of A: {a_mat.shape}")
print(f"Shape of B: {b_mat.shape}")
print(f"Shape of C = A.dot(B): {c_mat.shape}")
assert c_mat.shape == (2, 3), "Shape rule verified!"
```

## 5. Chapter Review & Synthesis Questions

### Conceptual Self-Check:
1. **Q: What is the physical / intuitive meaning of a negative weight?**  
   *A:* A negative weight represents negative correlation (inhibition). As the corresponding input increases, the network's prediction decreases.

2. **Q: Why does stacking two linear layers without an activation function fail to increase representational power?**  
   *A:* Because matrix multiplication is associative: $\mathbf{W}_2(\mathbf{W}_1 \mathbf{x}) = (\mathbf{W}_2 \mathbf{W}_1)\mathbf{x} = \mathbf{W}_{net}\mathbf{x}$. The product of two linear transformations is mathematically equivalent to a single linear transformation with weights $\mathbf{W}_{net} = \mathbf{W}_2 \mathbf{W}_1$.

3. **Q: How does the vector dot product act as a "similarity metric"?**  
   *A:* The dot product computes $\sum x_i w_i$. When the input vector features match the sign and relative scale of the weight vector, the terms multiply to large positive values, producing maximum excitation in the downstream node.
